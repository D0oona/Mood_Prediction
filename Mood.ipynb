```python
X_train,X_test,Y_train,Y_test= train_test_split(df.iloc[:,:-1],df.iloc[:,-1],test_size=0.2,random_state=42)
```

```python
X_train.head()
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.preprocessing import OrdinalEncoder
transformer = ColumnTransformer(
    transformers=[
        ('tnf1', OrdinalEncoder(
            categories=[['Low', 'Medium', 'High']]
        ), ['work_stress']),

        ('tnf2', OrdinalEncoder(
            categories=[['Light', 'Moderate', 'Heavy']]
        ), ['workload']),
        ('tnf3', OneHotEncoder(sparse_output=False,drop='first'), ['weather']),

        ('tnf4', OrdinalEncoder(
            categories=[['Poor', 'Fair', 'Good', 'Excellent']]
        ), ['sleep_quality'])
        
    ],
    remainder='passthrough'
)

```

```python
X_train_encoded = transformer.fit_transform(X_train)
X_test_encoded = transformer.transform(X_test)
```

```python
X_train_encoded.shape
```

```python
X_test_encoded.shape
```

```python
model = RandomForestClassifier(
    n_estimators=300,
    max_depth=6,
    min_samples_split=10,
    min_samples_leaf=8,
    max_features='sqrt',
    random_state=42
)

model.fit(X_train_encoded, Y_train)

print("Training accuracy:", model.score(X_train_encoded, Y_train))
print("Testing accuracy:", model.score(X_test_encoded, Y_test))
```

```python
importances = pd.Series(
    model.feature_importances_,
    index=transformer.get_feature_names_out()
)

print(importances.sort_values(ascending=False))
```
